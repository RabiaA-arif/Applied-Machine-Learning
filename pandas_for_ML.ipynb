## Lets Explore Panda

```python
import pandas as pd
import numpy as np
print(pd.__version__)
```

## Panda
     python library for working on tabular data

```python
person_info = {"name":"Miran","Age":26,"Profession":"student"}
# print(person_info["Age"])
print(person_info)
print(list(person_info.keys()))
```

```python
# Series : one dimensional label array
student = pd.Series([11,22,33,44,55],index=["ALi","Aliyan","Aliyar","Alim","Alina"], name="Age")
# print(student)
# converting panda series into numpy array
print("value:", student.to_numpy())
print("index:", list(student.index))
# accessing by label
print(student["ALi"])
# accessing by position
print(student.iloc[0])
```

```python
print(student + 5)  # adding 5 in all values
```

**DataFrame**

      multiple series form a dataframe

```python
data = {
    "Area":[1100,1200,1300,1400],
    "Rooms":[2,3,1,4],
    "Price":[100000,200000,30000,40000]
}
# print(data)
df = pd.DataFrame(data)
print(df)
print(df.shape)  # (row,column)
print(df.columns)  # names of column
print(df.index)  # index label
print(df.dtypes)  # datatypes of each columns
print(df.size)  # number of toatal cell
print(df.ndim)  # total dimension
print(df.to_numpy()) # converting numpy array
```
