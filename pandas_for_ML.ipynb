```python
# Series : one dimensional label array
student = pd.Series([11,22,33,44,55],index=["ALi","Aliyan","Aliyar","Alim","Alina"], name="Age")
# print(student)
# converting panda series into numpy array
print("value:", student.to_numpy())
print("index:", list(student.index))
# accessing by label
print(student["ALi"])
# accessing by position
print(student.iloc[0])
```

**DataFrame**

      multiple series form a dataframe

```python
data = {
    "Area":[1100,1200,1300,1400],
    "Rooms":[2,3,1,4],
    "Price":[100000,200000,30000,40000]
}
# print(data)
df = pd.DataFrame(data)
print(df)
print(df.shape)  # (row,column)
print(df.columns)  # names of column
print(df.index)  # index label
print(df.dtypes)  # datatypes of each columns
print(df.size)  # number of toatal cell
print(df.ndim)  # total dimension
print(df.to_numpy()) # converting numpy array
```

## Selecting Data
    using [],loc,iloc,boolean conditions and query
    loc uses names and iloc uses position

    df[''] : series
    df[['','']] : several columns
    df.loc[label]
    df.iloc[pos]
    df.loc[row, cols],df.iloc[r, c]
    df[mask] filter rowa
    & ,\ |,~ : combine condition
    df['col'].between(a,b) : range
    df.query('salary'> 50000) : sql style text

```python
emp = pd.DataFrame({
    "name":["Omar","Osman","Amaan","Ali","Aliyan","Assan"],
    "dept":["IT","HR","WSE","AIE","QA",'Manager'],
    "salary":[200000,230000,400000,520000,130000,None],
    "age":[21,25,23,25,22,33],
    "city":["Umerkot","Hydrabad","Nawabshah","Karachi","Lahore","Lahore"]

})
```

```python
print(emp['name'])
```

```python
print(emp.loc[1])   # row with label 1
```

```python
print(emp.iloc[1:3,[0,2,3]])
```

```python
print(emp.loc[1:3,['name']])
```

```python
print(emp.info())
```

```python
print(emp.isna().sum())
```

```python
high = emp['age'] > 21
print(high)
```

```python
print(emp[(emp['age'] > 21) & (emp['salary'] > 200000)])
```

```python
print(emp[emp['city'].isin(['Karachi'])][['name','city']])
```

```python
print(emp[emp['age'].isin(['23'])]['age'])
```
